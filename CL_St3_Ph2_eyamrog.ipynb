```python
import pandas as pd
import re

data = []

with open('your_file.txt', 'r', encoding='utf-8') as f:
    lines = [line.strip() for line in f if line.strip()]

i = 0
while i < len(lines):
    # Parse header
    header = re.split(r'\s+', lines[i])
    filename, type_token, word_length, word_count = header[0], float(header[1]), float(header[2]), int(header[3])
    i += 1

    # Parse 10 lines of 15 values each
    feature_values = []
    for _ in range(10):
        feature_values.extend([float(x) for x in re.split(r'\s+', lines[i])])
        i += 1

    # Parse final line: 5 known factors + 5 unknowns
    extra = [float(x) for x in re.split(r'\s+', lines[i])]
    i += 1

    entry = {
        'Filename': filename,
        'Type/Token': type_token,
        'Word Length': word_length,
        'Word Count': word_count,
        **{f'Feature {j+1}': val for j, val in enumerate(feature_values)},
        **{f'Factor {k+1} Score': extra[k] for k in range(5)},
        **{f'Unknown {k-4}' : extra[k] for k in range(5,10)}
    }

    data.append(entry)

# Create DataFrame
df_stats = pd.DataFrame(data)
```
