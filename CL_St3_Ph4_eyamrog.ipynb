```python
import pandas as pd
import re
import matplotlib.pyplot as plt
```

## Define input variables

```python
input_directory = 'cl_st3_ph1_eyamrog'
output_directory = 'cl_st3_ph1_eyamrog'
# EL2AP
counts_el2ap_1 = 'cl_st3_ph1_eyamrog/el2ap/el2ap_1/tagcount/counts_el2ap_1.txt'
counts_el2ap_2 = 'cl_st3_ph1_eyamrog/el2ap/el2ap_2/tagcount/counts_el2ap_2.txt'
counts_el2ap_3 = 'cl_st3_ph1_eyamrog/el2ap/el2ap_3/tagcount/counts_el2ap_3.txt'
counts_el2ap_4 = 'cl_st3_ph1_eyamrog/el2ap/el2ap_4/tagcount/counts_el2ap_4.txt'
counts_el2ap_5 = 'cl_st3_ph1_eyamrog/el2ap/el2ap_5/tagcount/counts_el2ap_5.txt'
counts_el2ap_6 = 'cl_st3_ph1_eyamrog/el2ap/el2ap_6/tagcount/counts_el2ap_6.txt'
counts_el2ap_7 = 'cl_st3_ph1_eyamrog/el2ap/el2ap_7/tagcount/counts_el2ap_7.txt'
# AI-EL2AP
counts_ai_el2ap_1 = 'cl_st3_ph1_eyamrog/ai-el2ap/ai-el2ap_1/tagcount/counts_ai-el2ap_1.txt'
counts_ai_el2ap_2 = 'cl_st3_ph1_eyamrog/ai-el2ap/ai-el2ap_2/tagcount/counts_ai-el2ap_2.txt'
counts_ai_el2ap_3 = 'cl_st3_ph1_eyamrog/ai-el2ap/ai-el2ap_3/tagcount/counts_ai-el2ap_3.txt'
counts_ai_el2ap_4 = 'cl_st3_ph1_eyamrog/ai-el2ap/ai-el2ap_4/tagcount/counts_ai-el2ap_4.txt'
counts_ai_el2ap_5 = 'cl_st3_ph1_eyamrog/ai-el2ap/ai-el2ap_5/tagcount/counts_ai-el2ap_5.txt'
counts_ai_el2ap_6 = 'cl_st3_ph1_eyamrog/ai-el2ap/ai-el2ap_6/tagcount/counts_ai-el2ap_6.txt'
counts_ai_el2ap_7 = 'cl_st3_ph1_eyamrog/ai-el2ap/ai-el2ap_7/tagcount/counts_ai-el2ap_7.txt'
```

## Extract tagcount summaries from the tagcount files

```python
def tagcount_summary(tagcount_file_path):
    '''
    Extract a summary of the tagcount files
    '''
    data = []

    with open(tagcount_file_path, 'r', encoding='utf-8') as f:
        lines = [line.strip() for line in f if line.strip()]

    i = 0
    while i < len(lines):
        # Parse first line (metadata)
        header = re.split(r'\s+', lines[i])
        filename = header[0]
        type_token = float(header[1])
        word_length = float(header[2])
        word_count = int(header[3])
        i += 1

        # Skip the 10 lines of feature counting
        i += 10

        # Parse final line: 10 values (first 5 = known factor scores, last 5 = unknown)
        tail_values = [float(val) for val in re.split(r'\s+', lines[i])]
        i += 1

        # Store as a dictionary
        entry = {
            'Filename': filename,
            'Type/Token': type_token,
            'Word Length': word_length,
            'Word Count': word_count,
            **{f'Factor {k+1} Score': tail_values[k] for k in range(5)},
            **{f'Unknown {k-4}': tail_values[k] for k in range(5, 10)}
        }
        data.append(entry)

    # Create a DataFrame
    df_tagcount_summary = pd.DataFrame(data)

    return df_tagcount_summary
```

## Import the data into a DataFrame

```python
df_el2ap_balanced = pd.read_json(f"{input_directory}/df_el2ap_balanced.jsonl", lines=True)
```

```python
df_el2ap_balanced['Submitted'] = pd.to_datetime(df_el2ap_balanced['Submitted'], unit='ms')
df_el2ap_balanced['Posted'] = pd.to_datetime(df_el2ap_balanced['Posted'], unit='ms')
```

### Add the `EL2AP Filename` and `AI-EL2AP Filename` columns

```python
df_el2ap_balanced['EL2AP Filename'] = (
    df_el2ap_balanced['Text ID'].astype(str) + '_' +
    df_el2ap_balanced['Section Code'].astype(str) + '_' +
    df_el2ap_balanced['Paragraph Code'].astype(str) + '_el2ap.txt'
)
```

```python
df_el2ap_balanced['AI-EL2AP Filename'] = (
    df_el2ap_balanced['Text ID'].astype(str) + '_' +
    df_el2ap_balanced['Section Code'].astype(str) + '_' +
    df_el2ap_balanced['Paragraph Code'].astype(str) + '_ai-el2ap.txt'
)
```

```python
df_el2ap_balanced.shape
```

## Splitting the `df_el2ap_balanced` corpus into `df_el2ap_dimensions` and `df_ai_el2ap_dimensions` corpora

```python
df_el2ap_dimensions = df_el2ap_balanced
```

```python
df_ai_el2ap_dimensions = df_el2ap_balanced
```

## Enriching `df_el2ap_dimensions` with factor scores

### Consolidating the tagcount summaries for `EL2AP`

```python
df_tagcount_el2ap_1 = tagcount_summary(counts_el2ap_1)
df_tagcount_el2ap_2 = tagcount_summary(counts_el2ap_2)
df_tagcount_el2ap_3 = tagcount_summary(counts_el2ap_3)
df_tagcount_el2ap_4 = tagcount_summary(counts_el2ap_4)
df_tagcount_el2ap_5 = tagcount_summary(counts_el2ap_5)
df_tagcount_el2ap_6 = tagcount_summary(counts_el2ap_6)
df_tagcount_el2ap_7 = tagcount_summary(counts_el2ap_7)

df_tagcount_el2ap = pd.concat([
    df_tagcount_el2ap_1,
    df_tagcount_el2ap_2,
    df_tagcount_el2ap_3,
    df_tagcount_el2ap_4,
    df_tagcount_el2ap_5,
    df_tagcount_el2ap_6,
    df_tagcount_el2ap_7
], ignore_index=True)
```

```python
df_tagcount_el2ap.drop(columns=['Unknown 1', 'Unknown 2', 'Unknown 3', 'Unknown 4', 'Unknown 5'], inplace=True)
```

```python
df_tagcount_el2ap.rename(columns={'Filename': 'EL2AP Filename'}, inplace=True)
```

```python
df_tagcount_el2ap.shape
```

### Removing texts that have been excluded by `TagCount`

The `TagCount` processing excluded a few texts that have not scored in any of the dimensions. They should removed from the corpus.

```python
tagcount_el2ap_delta = df_el2ap_balanced[
    ~df_el2ap_balanced['EL2AP Filename'].isin(df_tagcount_el2ap['EL2AP Filename'])
]['EL2AP Filename'].tolist()
```

```python
len(tagcount_el2ap_delta)
```

```python
tagcount_el2ap_delta
```

```python
df_el2ap_dimensions = df_el2ap_dimensions[~df_el2ap_dimensions['EL2AP Filename'].isin(tagcount_el2ap_delta)]
```

```python
df_el2ap_dimensions.shape
```

### Merging `df_tagcount_el2ap` into `df_el2ap_dimensions` by the `EL2AP Filename` column

```python
df_el2ap_dimensions = df_el2ap_dimensions.merge(df_tagcount_el2ap, on='EL2AP Filename', how='left')
```

```python
df_el2ap_dimensions
```

```python
df_el2ap_dimensions.isna().sum()
```

#### Exporting to a file

```python
df_el2ap_dimensions.to_json(f"{output_directory}/df_el2ap_dimensions.jsonl", orient='records', lines=True)
```
